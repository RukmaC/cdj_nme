```python
#A1 use np.arange to create a new list, called even_list_1 that contains all the even numbers from 0 to 100 (exclusive).
even_list_1 = np.arange(0, 100, 2)
print(even_list_1)
```

```python
#the bracket [] allows us to access a specific element in the list.
print(even_list_1[0]) #prints the first element in the list

#use the : to access elements accross a certain range 
print(even_list_1[5:15]) #prints the first 10 elements in the list

#: can also be used to access elements from the beginning or end of a list.
print(random_list[:5]) #prints the first 5 elements in the list
print(random_list[90 :]) #prints the last 10 elements in the list
```

```python
#A2 

#print the element at the 90th index 
print(random_list[90])
#ninety_index = random_list[90] 
#good practice to not put numericals in variable names - write out ninety

#print the elements from index 10 to index 21
print(random_list[10:22])

#print the last 10 elemeents of the array 
print(random_list[-10:])
```

```python
#doing maf operations on the list 
#want the list to start at 1 instead of 0 
random_list = random_list + 1
#if u run this a lot then it would keep adding 1 - make you actually want to reassign variable 
#add 1 to all elements in the list
print(random_list)
```

```python
#boolean indexing can also be performed element-wise in an array 
#boolean index array that shows where all the odds(FALSE) are: 
print(random_list % 2== 0)

even_list_2 = random_list[random_list%2==0]
print(even_list_2)
```

```python
#initialize an empty array 
even_list_3 = np.array([])
#for i in range just means for each value i 

for i in range(0,101):
    #check is the value i is even
    if i % 2 ==0:
        #use the np.append i since it's even 
        even_list_3 = np.append(arr=even_list_3, values= i)
print(even_list_3)
```

```python
#make an array of nums 0-200 (exclusive) 
big_list = np. arange(200)

#multiply numbers resulting in a list 
even_list_4 = [2*x for x in big_list if x<=50]

even_list_4 = np.array(even_list_4)
print(even_list_4)

#these all show that there are different methods to get the same result - just depends on what you want to do and how you want to do it
```
